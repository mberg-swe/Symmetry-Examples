### SMSA 17.1 — Repetition of graphical calculus

Source: SMSA 7.8, 2026-11-05, `smsa.pdf`. References below use section and equation numbers. Read the corresponding SMSA passage first. Native DisCoPy drawings run **top to bottom**: this preview reverses SMSA’s usual vertical flow while preserving left-to-right tensor order. A coupon replaces SMSA’s specialized glyph where necessary. This notebook uses no user-defined functions or assertion checks.

Equations (17.4), (17.6), (17.8)–(17.9): `f >> h` translates $h\circ f$, and `@` translates $\otimes$. The equality below is checked by DisCoPy’s monoidal normalization; coupons stand for the unspecified maps in SMSA.

```python
from discopy import monoidal
X, Y, Z = map(monoidal.Ty, ['X', 'Y', 'Z'])
f = monoidal.Box('f', X, Y)
h = monoidal.Box('h', Y, Z)
fprime = monoidal.Box("f'", X, Y)
hprime = monoidal.Box("h'", Y, Z)
lhs = (f @ fprime) >> (h @ hprime)
rhs = (f >> h) @ (fprime >> hprime)
print('17.8 left:', lhs)
print('17.8 right:', rhs)
print('Left normal form:', lhs.normal_form())
print('Right normal form:', rhs.normal_form())
print('Same normal form:', lhs.normal_form() == rhs.normal_form())
```
