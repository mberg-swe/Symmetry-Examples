### SMSA 3.4 — Dictionary of native generators

**Source:** SMSA 7.8, 2026-11-05, §3.4 summary of graphical calculus. Local source: `smsa.pdf`. Source SHA-256: `6922b174be6719adbe0bfa0573cce44f0e5a7a54c4fafd5e43969c952813b8fe`.

**Translation conventions.** SMSA reads diagrams bottom to top. These unmodified native DisCoPy drawings read top to bottom: inputs are above outputs. Tensor factors retain their left-to-right order. `f >> g` means SMSA's $g\circ f$; `f @ g` means $f\otimes g$. The empty type represents the invisible unit wire. Coupons stand for the named SMSA maps even when SMSA uses a branching vertex. No custom functions or rewrite rules are introduced.

Checks print their computed objects and comparison results and continue when a mathematical comparison is false. A comparison in a concrete tensor model establishes only that model instance.

The table in SMSA becomes a list of typed native coupons. This is a syntax dictionary: declaring a coupon does not impose the algebra, coalgebra or antipode equations. Those equations are evaluated in the neighboring notebooks. Unit and counit have an empty type on one side, representing the invisible unit wire.

```python
from discopy import symmetric
A = symmetric.Ty("A")
unit = symmetric.Ty()
mu = symmetric.Box("mu", A @ A, A)
eta = symmetric.Box("eta", unit, A)
Delta = symmetric.Box("Delta", A, A @ A)
epsilon = symmetric.Box("epsilon", A, unit)
s = symmetric.Box("s", A, A)
```

```python
for diagram in [mu, eta, Delta, epsilon, s]:
    print(str(diagram), ':', repr(diagram.dom), '->', repr(diagram.cod))
    diagram.draw(figsize=(3, 2))
```

**Rendering choice.** Multiplication, comultiplication, unit and counit appear as labeled boxes, whereas SMSA often uses vertices. Using DisCoPy Frobenius spiders here would impose additional algebraic structure; these general typed boxes preserve the intended scope.
